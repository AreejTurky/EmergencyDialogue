```python
prompt_template="أعطني ملخص لهذه المكالمة على شكل استمارة:\n"
```

```python
annote_result=[]
call=[]
```

```python
samples.iloc[0]
```

```python
for i in range(len(data['conv'])):
    response = co.generate(
      model='command-r-plus',
      prompt = f"{prompt_template} {data['conv'][i]}",
      max_tokens=300,
      temperature=0.9,
      k=0,
      stop_sequences=[],
      return_likelihoods='NONE')
    
    annote_result.append(response.generations[0].text)
    call.append(data['conv'][i])
```
